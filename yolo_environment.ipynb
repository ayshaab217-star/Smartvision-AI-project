```python
import os

labels_path = r"D:\Smartvision_project\smartvision_dataset\detection\labels"

class_ids = set()
invalid_lines = []
total_boxes = 0

for file in os.listdir(labels_path):
    if not file.endswith(".txt"):
        continue

    path = os.path.join(labels_path, file)

    with open(path, "r") as f:
        for line_number, line in enumerate(f, start=1):
            parts = line.strip().split()

            if not parts:
                continue

            total_boxes += 1

            try:
                class_id = int(parts[0])
                class_ids.add(class_id)

                if class_id < 0 or class_id > 24:
                    invalid_lines.append(
                        (file, line_number, class_id)
                    )

            except ValueError:
                invalid_lines.append(
                    (file, line_number, "Invalid class ID")
                )

print("Unique class IDs found:", sorted(class_ids))
print("Total bounding boxes:", total_boxes)
print("Invalid class IDs:", len(invalid_lines))

if invalid_lines:
    print("\nFirst invalid entries:")
    for item in invalid_lines[:10]:
        print(item)
else:
    print("\n✅ All class IDs are valid (0-24)")
```

```python
import os

train_images = os.path.join(detection_path, "train", "images")
train_labels = os.path.join(detection_path, "train", "labels")

val_images = os.path.join(detection_path, "val", "images")
val_labels = os.path.join(detection_path, "val", "labels")

def count_files(folder, extensions):
    if not os.path.exists(folder):
        return 0
    return sum(
        1 for f in os.listdir(folder)
        if f.lower().endswith(extensions)
    )

print("TRAIN")
print("Images:", count_files(train_images, (".jpg", ".jpeg", ".png")))
print("Labels:", count_files(train_labels, (".txt",)))

print("\nVALIDATION")
print("Images:", count_files(val_images, (".jpg", ".jpeg", ".png")))
print("Labels:", count_files(val_labels, (".txt",)))
```

```python
import os
import shutil

# Paths
source_labels = os.path.join(detection_path, "labels")
train_labels = os.path.join(detection_path, "train", "labels")

# Create train/labels folder
os.makedirs(train_labels, exist_ok=True)

# Copy labels matching the training images
train_image_files = [
    f for f in os.listdir(train_images)
    if f.lower().endswith((".jpg", ".jpeg", ".png"))
]

copied = 0
missing = 0

for image_file in train_image_files:
    label_file = os.path.splitext(image_file)[0] + ".txt"
    
    source = os.path.join(source_labels, label_file)
    destination = os.path.join(train_labels, label_file)
    
    if os.path.exists(source):
        shutil.copy2(source, destination)
        copied += 1
    else:
        missing += 1

print("✅ Training labels created")
print("Training images:", len(train_image_files))
print("Labels copied:", copied)
print("Missing labels:", missing)
```

```python
import os

def check_matching_files(images_folder, labels_folder):
    images = {
        os.path.splitext(f)[0]
        for f in os.listdir(images_folder)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    }

    labels = {
        os.path.splitext(f)[0]
        for f in os.listdir(labels_folder)
        if f.lower().endswith(".txt")
    }

    missing_labels = images - labels
    extra_labels = labels - images

    return images, labels, missing_labels, extra_labels


# Check TRAIN
train_imgs, train_lbls, train_missing, train_extra = check_matching_files(
    train_images,
    train_labels
)

# Check VALIDATION
val_imgs, val_lbls, val_missing, val_extra = check_matching_files(
    val_images,
    val_labels
)

print("========== FINAL DATASET CHECK ==========")

print("\nTRAIN")
print("Images:", len(train_imgs))
print("Labels:", len(train_lbls))
print("Missing labels:", len(train_missing))
print("Extra labels:", len(train_extra))

print("\nVALIDATION")
print("Images:", len(val_imgs))
print("Labels:", len(val_lbls))
print("Missing labels:", len(val_missing))
print("Extra labels:", len(val_extra))

if (
    len(train_missing) == 0
    and len(train_extra) == 0
    and len(val_missing) == 0
    and len(val_extra) == 0
):
    print("\n🎉 YOLO DATASET IS READY FOR TRAINING!")
else:
    print("\n⚠️ Dataset still needs checking.")
```
